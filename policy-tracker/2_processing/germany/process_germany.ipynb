# processing german bgbl raw data



```python
import json


file_path = "S:/stbuchho/crises_and_welfare_states/policy-tracker/data/raw/germany/bgbl1_2019-2022_20260408.json"

with open(file_path, "r", encoding="utf-8") as file:
    data = json.load(file)

entries = data if isinstance(data, list) else list(data.values())
for i, entry in enumerate(entries[:70]):
    print(f"--- Entry {i+1} ---")
    print(json.dumps(entry, ensure_ascii=False, indent=2))


```
