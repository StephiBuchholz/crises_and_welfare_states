```python
#schema for preliminary policy link collection

schema = {
    "title": str,
    "url": str,
    "scope": str, #only use ["international", "eu", "germany", "FillInCountry"] #always lower-case
    "A_or_B": ["A", "B", "both"],
    "topics": list,
    "type": ["tracker", "archive", "API", "database"], #tracker: policy evolution over time, oft. incl. indices, visuals, codes; archive: data collection focused on gathering
    "notes": str,
    "timeframe": list, #use NA if not clear, use "0" if decades long collection or when time-span together with the start and end year
    "found": ["own_research_any", "OxSupTrack", "FalisseMcAteer2022", "own_research_ILO", "own_research_EU"], #if adding categories, add to validation function, too // own_research_any for entries found 'randomly'
    "final_judge": ["fit", "nonfit", "pending"] #this is a preliminary judgement, fit may end up unfit after full data exploration 
}




#validator for my entries in the link collection (site links)
##checks if all variables are present and if type of variable and, if applicable, its categories, are correct

def validate_source_entry(entry):
    errors = []

    if "title" not in entry or not isinstance(entry["title"], str) or not entry["title"].strip():
        errors.append("Missing or invalid 'title'")
    
    if "url" not in entry or not isinstance(entry["url"], str) or not entry["url"].startswith("http"):
        errors.append("Missing or invalid 'url'")
    
    if "scope" not in entry or not isinstance(entry["scope"], str):
        errors.append("Missing or invalid 'scope'")

    if "A_or_B" not in entry or not isinstance(entry["A_or_B"], str):
        errors.append("Missing or invalid 'A_or_B'")

    if "topics" not in entry or not isinstance(entry["topics"], list):
        errors.append("Missing or invalid 'topics'")
    
    if "type" not in entry or entry["type"] not in ["tracker", "archive", "API", "database"]:
        errors.append("Invalid or missing 'type' (must be 'tracker', 'archive', 'database' or 'API')")
    
    if "timeframe" not in entry or not isinstance(entry["timeframe"], list):
        errors.append("Missing or invalid 'timeframe' (must be a list)")

    if "found" not in entry or entry["found"] not in ["own_research", "OxSupTrack", "FalisseMcAteer2022", "own_research_ILO", "own_research_EU"]:
        errors.append("Invalid or missing 'found' (must be 'own_research', 'OxSupTrack', 'FalisseMcAteer2022', 'own_research_ILO', 'own_research_EU)")
    
    if "final_judge" not in entry or entry["final_judge"] not in ["fit", "nonfit", "pending"]:
        errors.append("Invalid or missing 'final_judge' (must be 'fit', 'nonfit' or 'pending')")

    return errors

#code for running validation check below site_link list ->

```
