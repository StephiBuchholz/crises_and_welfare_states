```python
from pathlib import Path
from datetime import date
import pandas as pd
from keybert import KeyBERT # before: pip install keybert 
from yake import KeywordExtractor #before: pip install yake
from collections import Counter
```

## 1. import and parse naive search results

```python
input_dir = Path("../02_processed")  # reads scopus/ and webofscience/ subfolders

all_dfs = []

# defining desired output columns (they equal those in the scopus files)
scopus_cols = [
    "title", "abstract", "eid", "authors", "journal", "issue",
    "volume", "keywords", "doctype", "year", "doi",
    "pagestart", "pageend"
]

# macthing current to desired column names for wos files
wos_to_scopus = {
    "Article Title": "title",
    "Abstract": "abstract",
    "UT (Unique WOS ID)": "eid",
    "Authors": "authors",
    "Source Title": "journal",
    "Issue": "issue",
    "Volume": "volume",
    "Author Keywords": "keywords",
    "Document Type": "doctype",
    "Publication Year": "year",
    "DOI": "doi",
    "Start Page": "pagestart",
    "End Page": "pageend",
}

# loop over scopus files (only *_adapt_columns.csv files from Stage 1)
for csv_path in (input_dir / "scopus").glob("*_adapt_columns.csv"):
    df = pd.read_csv(csv_path)
    
    # normalize column names
    df.columns = df.columns.str.lower()

    # keep only desired columns
    df = df[[c for c in scopus_cols if c in df.columns]]
    
    #create column for "database" (scopus or wos)
    df["database"] = "scopus"

    # add a column "search run" (terms used)
    name= csv_path.stem
    search_run = name.replace("scopus_", "").split("_", 1)[1]
    search_run = search_run.replace("_adapt_columns", "")
    df["search_run"]=search_run
    
    #add a column "title_normalized"
    df["title_normalized"]= df["title"].str.lower().str.strip()

    #add a column "abs_normalized"
    df["abs_normalized"]= df["abstract"].str.lower().str.strip()
    
    all_dfs.append(df)

# loop over wos files
for csv_path in (input_dir / "webofscience").glob("*.csv"):
    df = pd.read_csv(csv_path)
    
    # rename wos columns to scopus schema
    df = df.rename(columns=wos_to_scopus)

    # keep only mapped columns
    df = df[[c for c in scopus_cols if c in df.columns]]

    #add to column for "database" (scopus or wos)
    df["database"] = "wos"

    #add to column "search_run"
    name= csv_path.stem
    search_run = name.replace("wos_", "").split("_", 1)[1]
    df["search_run"]=search_run

    #add a column "title_normalized"
    df["title_normalized"]= df["title"].str.lower().str.strip()

    #add a column "abs_normalized"
    df["abs_normalized"]= df["abstract"].str.lower().str.strip()
    
    all_dfs.append(df)

# compile to one dataframe
allresults_df = pd.concat(all_dfs, ignore_index=True)

print(f"Combined dataframe shape: {allresults_df.shape}")
allresults_df.head(100) #show part of df
```

```python
categories = allresults_df["search_run"].unique()
print(categories)
```

## 2. deduplicate and combine to df

### 2.1 deduplication



```python
# count entries before deduplication
print(f"before deduplication, the dataframe allresults_df contains {len(allresults_df)} entries.")

# 1st deduplicate, by doi
## count number of missings in column "doi"
print(f"out of {len(allresults_df)} entries, there are {allresults_df['doi'].isna().sum()} entries with the doi missing.")

##using doi as anchor for deduplication
### deduplicate rows with a DOI
dedup_with_doi = (
    allresults_df[allresults_df['doi'].notna()]
    .assign(is_globalsouth=allresults_df['search_run'].str.startswith("globalsouth_", na=False)) #prioritizes entries from search_runs on "globalsouth_"; i later want to be able to analyse them separately and therefore do not want this easy identifier dropped in deduplication (rather drop their duplicates from other search runs)
    .sort_values('is_globalsouth', ascending=False)
    .drop_duplicates(subset='doi', keep='first')
)

### keep rows with missing DOI
missing_doi_df = allresults_df[allresults_df['doi'].isna()]

### combine both
unique_results_df = pd.concat([dedup_with_doi, missing_doi_df], ignore_index=True)

### print counts
print(f"after deduplication, unique_results_df contains {len(unique_results_df)} entries, including {unique_results_df['doi'].isna().sum()} with missing doi.")
```

```python
# 2nd deduplication, by title
## check for duplicate titles in unique_results_df
duplicate_titles = unique_results_df[unique_results_df.duplicated(subset='title_normalized', keep=False)]
print(f"number of entries with duplicate titles: {len(duplicate_titles)}")
if not duplicate_titles.empty:
    print("duplicate titles and their counts:")
    print(duplicate_titles.groupby('title').size().sort_values(ascending=False))

duplicate_titles.head()

## deduplicate by title 
### sort so rows with a doi come first within each title
unique_results_tmp = unique_results_df.copy()
unique_results_tmp['_has_doi'] = unique_results_tmp['doi'].notna()

### deduplicate by title, keeping rows with doi preferentially
final_results_df = (
    unique_results_tmp.sort_values(['_has_doi', 'is_globalsouth'], ascending=False)
       .drop_duplicates(subset='title_normalized', keep='first')
       .drop(columns='_has_doi')
       .reset_index(drop=True)
)


print(f"the final dataframe final_results_df has been deduplicated by title_normalized and doi, and has {len(final_results_df)} entries, including {final_results_df['doi'].isna().sum()} with missing doi.")
del unique_results_tmp
```

### 2.2 separate "welfare state change" literature

this step is domain specific and specific to my naive search process since i included searches regarding welfare state change. however, this is a separate part of my literature review and requires a separate search strategy alltogether since the naive results likely only contain empirical studies, but not necessarily the foundational publications on keywords like retrenchment, social learning etc.. i therefore separate the results of that search into a new dataframe and delete them from final_results_df.

```python
# create subset of results that excludes results stemming from the search term "welfarestatechange" (side topic)

# Create a new column 'is_wschange' 
## 1 if 'welfarestatechange' appears in 'search_run', 0 otherwise
final_results_df['is_wschange'] = final_results_df['search_run'].str.contains( #str.contains(checks a field and returns boolean)
    'welfarestatechange', case=False, na=False #case for lower/upper case sensitivity / NA treated as false
).astype(int) #true/false expressed as 1/0

##count how many entries stem from welfarestate change as search term
final_results_df['is_wschange'].value_counts()

## new dataframe with only welfare state change entries
finres_wschange=final_results_df[final_results_df["is_wschange"]==1].copy()

## drop rows with welfare state change from final_results_df

final_results_df=final_results_df[final_results_df["is_wschange"]!=1]


print(f"the final dataframe final_results_df has {len(final_results_df)} entries after the exclusion of entries from the search runs on welfarestate change, these were {len(finres_wschange)}")
```

```python
#drop column "is_wschange" in final_results_df

final_results_df=final_results_df.drop(columns=["is_wschange"])
```

## 3. key word extraction

in this section, two keyword extraction methods, keyBERT and yake, are applied to the title and abstract of each entry in the dataframe final_results_df separately. the resulting 5 most relevant keywords/keyphrases are stored in new columns "keybird" and "yake". the keyword come in tuples with their score (document-interal relevance). for score interpretation see descriptions in 3.1 and 3.2, respectively.

in a next step, these keywords, for keybird and yake, respectively, are combined over all entries into the most relevant keywords alltogether simply on the basis of highest frequency.

## 3.1 keyBERT
link: https://github.com/MaartenGr/KeyBERT?tab=readme-ov-file#llms

the first out of two keyword extraction methods used is keyBERT. it uitlizes "BERT-embeddings and simple cosine simiarity to find the subphrased in a document that are the most simiar to the document itself. first, it extracts document-level embeddings to get a document-level representation. then, word embeddings are extracted for N-gram words/phrases. cosine similarity is used to find words/phrases that are most similar to the document."

- score interpretation: higher score means higher representation of overall document/text

- citation bibtex:
@misc{grootendorst2020keybert,
  author       = {Maarten Grootendorst},
  title        = {KeyBERT: Minimal keyword extraction with BERT.},
  year         = 2020,
  publisher    = {Zenodo},
  version      = {v0.3.0},
  doi          = {10.5281/zenodo.4461265},
  url          = {https://doi.org/10.5281/zenodo.4461265}
}


```python
# build new variable that combines title and abstract into one (for keyword extraction)

final_results_df["title_abstract"] = (
    final_results_df["title_normalized"].fillna("") + " " +
    final_results_df["abstract"].fillna("")
)

# count missings in "title_abstract" (should be 0, every entry has at least a title)
print(f"there are {final_results_df['abstract'].isna().sum()} abstracts missing in final_results_df, so for them key word extraction will be based on the title only. {final_results_df['title_abstract'].isna().sum()} missings in title") 


# initialize keybert model
model = KeyBERT('distilbert-base-nli-mean-tokens')


#apply keyword extraction on every row separately for column "title_abstract"
final_results_df["keybert"] = final_results_df['title_abstract'].apply( #.apply() applies the function lamda on an all cells in title_abstracts
    lambda text: model.extract_keywords(text, keyphrase_ngram_range=(1,1)) if isinstance(text, str) and text.strip() else [] #keyphrase range set to (1,1), meaning only one-word phrases. for 2-word phrases, set to (1,2)
) # only applied when (if is instance...else) cell content is a string object, otherwise just fills in nothing


# quick sanity check
final_results_df[["title_normalized", "abstract", "keybert"]].head(100)
```

```python
# most common keywords over all final_results_df (keybert)

#flatten all keywords, ignore score by taking kw out of tuple into new object
all_kw_keybert=[kw for row in final_results_df["keybert"] for kw, score in row]

#count frequencies
keyword_counts_k=Counter(all_kw_keybert) #from collections library

#get most common-50 and -20 words

top_50_keybert= [kw for kw, count in keyword_counts_k.most_common(50)]
top_20_keybert= [kw for kw, count in keyword_counts_k.most_common(20)]

print(top_50_keybert)
print(top_20_keybert)
```

## 3.2 yake

link: https://github.com/INESCTEC/yake

the second keyword extraction method used is yake. it "is a lightweight unsupervised automatic keyword extraction method that uses text statistical features to select the most important keywords from a document"

- score interpretation: the lower (!) the score, the more relevant the keyword.

- citation bibtex: @InProceedings{10.1007/978-3-319-76941-7_80,
author="Campos, Ricardo
and Mangaravite, V{\'i}tor
and Pasquali, Arian
and Jorge, Al{\'i}pio M{\'a}rio
and Nunes, C{\'e}lia
and Jatowt, Adam",
editor="Pasi, Gabriella
and Piwowarski, Benjamin
and Azzopardi, Leif
and Hanbury, Allan",
title="YAKE! Collection-Independent Automatic Keyword Extractor",
booktitle="Advances in Information Retrieval",
year="2018",
publisher="Springer International Publishing",
address="Cham",
pages="806--810"
}


```python
# create instance of keyword extractor

yake_extract =KeywordExtractor(top=5, n=3) #customized to top-5 keywords, n is ngram size set to 3 (3-word phrases, default))

#apply keyword extraction on every row separately for column "title_abstract"
final_results_df["yake"] = final_results_df["title_abstract"].apply(
    lambda text: yake_extract.extract_keywords(text) if isinstance(text, str) and text.strip() else [] 
)

#quick sanity check
final_results_df[["title", "abstract", "keybert", "yake"]].head(100)
```

```python
# most common keywords over all final_results_df (yake)

#flatten all keywords, ignore score by taking kw out of tuple into new object
all_kw_yake=[kw for row in final_results_df["yake"] for kw, score in row]

#count frequencies
keyword_counts_y=Counter(all_kw_yake) #from collections library

#get most common-50 and -20 words

top_70_yake= [kw for kw, count in keyword_counts_y.most_common(70)]
top_20_yake= [kw for kw, count in keyword_counts_y.most_common(20)]

print(top_70_yake)
print(top_20_yake)
```
